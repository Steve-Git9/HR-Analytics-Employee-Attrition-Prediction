# IBM HR Attrition Dataset - Attribute Descriptions

---

## Target Variable

* **Attrition**: Indicates whether the employee has left the company.

  * `Yes`: the employee has left
  * `No`: the employee is still employed

---

## Employee Demographics

* **Age**: Employee age in years.
* **Gender**: Employee gender (`Male`, `Female`).
* **MaritalStatus**: Marital status (`Single`, `Married`, `Divorced`).
* **Education**: Education level (1: Below College, 2: College, 3: Bachelor, 4: Master, 5: Doctor).
* **EducationField**: Field of study (`Life Sciences`, `Medical`, `Marketing`, `Technical Degree`, etc.).

---

## Job and Organizational Information

* **Department**: Department of the employee (`Sales`, `Research & Development`, `Human Resources`).
* **JobRole**: Specific job role (`Sales Executive`, `Research Scientist`, `Laboratory Technician`, etc.).
* **JobLevel**: Hierarchical level in the organization (1–5, higher value indicates more senior position).
* **JobInvolvement**: Degree of involvement with the job (1: Low, 2: Medium, 3: High, 4: Very High).
* **BusinessTravel**: Frequency of business travel (`Non-Travel`, `Travel_Rarely`, `Travel_Frequently`).

---

## Compensation and Benefits

* **MonthlyIncome**: Actual monthly salary received by the employee.
* **HourlyRate**: Nominal hourly rate (does not reflect actual salary or hours worked).
* **DailyRate**: Nominal daily rate (not actual income).
* **MonthlyRate**: Nominal monthly rate (distinct from MonthlyIncome, weakly correlated with real salary).
* **StockOptionLevel**: Level of stock options granted (0–3).
* **PercentSalaryHike**: Percentage increase in the employee's salary during the last review. Indicates company recognition of performance and may influence retention.

---

## Satisfaction and Performance Metrics (Ordinal)

* **JobSatisfaction**: Job satisfaction (1: Low, 2: Medium, 3: High, 4: Very High).
* **EnvironmentSatisfaction**: Satisfaction with work environment (1–4).
* **RelationshipSatisfaction**: Satisfaction with workplace relationships (1–4).
* **WorkLifeBalance**: Perceived work-life balance (1: Bad, 2: Good, 3: Better, 4: Best).
* **PerformanceRating**: Performance evaluation (3: Good, 4: Excellent).

---

## Career History and Workload

* **NumCompaniesWorked**: Number of previous companies the employee has worked for. Indicates career mobility and experience across organizations.
* **TotalWorkingYears**: Total number of years the employee has worked in their professional career, across all companies.
* **YearsAtCompany**: Number of years the employee has worked at the company.
* **YearsInCurrentRole**: Number of years in the current role.
* **YearsSinceLastPromotion**: Years since the last promotion.
* **YearsWithCurrManager**: Years with the current manager.
* **TrainingTimesLastYear**: Number of training sessions attended in the previous year.
* **DistanceFromHome**: Distance from home to work (arbitrary units).
* **OverTime**: Indicates whether the employee works overtime (`Yes`, `No`).

---

## Technical Variables

* **EmployeeNumber**: Unique identifier for the employee (non-predictive).
* **EmployeeCount**: Constant variable (non-informative).
* **Over18**: Indicates whether the employee is over 18.
* **StandardHours**: Standard number of working hours.
